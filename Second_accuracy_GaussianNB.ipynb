```python
import pandas as pd
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
data = pd.read_excel("iris_train.xlsx", engine="openpyxl")
```

```python
data.head()
```

```python
x = data.iloc[:,:-1]
```

```python
y = data.iloc[:,-1]
```

```python
y
```

```python
x
```

```python
model = GaussianNB()
```

```python
model.fit(x.values,y.values)
```

```python
predictions = model.predict([[7	,3.2	,4.7	,1.4]])
```

```python
predictions
```

```python
predictions = model.predict([[6.8	,3.2	,5.9	,2.3]])
```

```python
predictions
```

```python
data_test = pd.read_excel("iris_test.xlsx", engine="openpyxl")
```

```python
data_test.head()
```

```python
x_test = data_test.iloc[:,:-1]
```

```python
predictions = model.predict(x_test.values)
```

```python
predictions
```

```python
y_test = data_test.iloc[:,-1]
```

```python
y_test.values
```

```python
actual = y_test.values
```

```python
count = 0
for i in range (0,len(actual)):
    if actual[i]==predictions[i]:
        count = count+1
print((count*100)/len(actual))
```
